Data Cleaning

```python
df_clean = df.copy()

print("Original Shape:", df.shape)
print("Cleaning Shape:", df_clean.shape)
```

```python
#Missing values we will find here

missing = df_clean.isnull().sum()

missing = missing[missing > 0].sort_values(ascending=False)

print("Columns with missing values:")
display(missing)
```

```python
#now we will find  the duplicate rows for this
duplicates = df_clean.duplicated().sum()
print("Total Duplicated Rows:", duplicates)
```

```python
df_clean.dtypes
```

```python
# Convert OrderDate into proper datetime format

df_clean["OrderDate"] = pd.to_datetime(df_clean["OrderDate"], errors="coerce")

print("OrderDate Data Type:")
print(df_clean["OrderDate"].dtype)
```

```python
print("Invalid/Missing Dates:", df_clean["OrderDate"].isnull().sum())
```

```python
text_columns = df_clean.select_dtypes(include="object").columns

for col in text_columns:
    df_clean[col] = df_clean[col].str.strip()
```

```python
print("Categories:")
print(df_clean["Category"].unique())

print("\nPayment Methods:")
print(df_clean["PaymentMethod"].unique())

print("\nOrder Status:")
print(df_clean["OrderStatus"].unique())
```

```python
# Identify numeric columns
numeric_cols = df_clean.select_dtypes(include=["int64", "float64"]).columns

print("Numeric Columns:")
print(numeric_cols.tolist())

print("\nMinimum Values:")
display(df_clean[numeric_cols].min())

print("\nMaximum Values:")
display(df_clean[numeric_cols].max())
```

```python
# Check negative values in numeric columns

for col in numeric_cols:
    negative_count = (df_clean[col] < 0).sum()
    print(f"{col}: {negative_count} negative values")
```

```python
print("FINAL DATA QUALITY CHECK")

print("\n1. Dataset Shape:")
print(df_clean.shape)

print("\n2. Missing Values:")
print(df_clean.isnull().sum().sum())

print("\n3. Duplicate Rows:")
print(df_clean.duplicated().sum())

print("\n4. Data Types:")
print(df_clean.dtypes)
```
